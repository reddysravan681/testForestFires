```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("Algerian_forest_fires_dataset.csv")
```

```python
df.head()
```

```python
##we are gonna predict temperature
df.info()
```

```python
#Data cleaning
df[df.isnull().any(axis=1)]
```

```python
df.loc[:122,"region"]=0
df.loc[122:,"region"]=1
```

```python
df.info()
```

```python
df.head()
```

```python
df['region']=df['region'].astype(int)
```

```python
df.isnull().sum()
```

```python
df=df.dropna().reset_index(drop=True)
```

```python
df.head()
```

```python
df.iloc[[122]]
```

```python
df=df.drop(122).reset_index(drop=True)
```

```python
df.head()
```

```python
df.iloc[[122]]
```

```python
df.columns
```

```python
#Removing spaces in column names
df.columns=df.columns.str.strip()
```

```python
df.columns
```

```python
df.info()
```

```python
df.columns
```

```python
#changing the required columns into integers
df[['day', 'month', 'year', 'Temperature', 'RH', 'Ws']]=df[['day', 'month', 'year', 'Temperature', 'RH', 'Ws']].astype(int)
```

```python
df.info()
```

```python
#changing other columns into float
remaining_columns=[feature for feature in df.columns if df[feature].dtypes=='str']
```

```python
remaining_columns
```

```python
for i in remaining_columns:
    if i!='Classes':
        df[i]=df[i].astype(float)
```

```python
df.info()
```

```python
df.describe()
```

```python
df.head()
```

```python
df.to_csv('Algerian_forest_fires_cleaned_dataset.csv',index=False)
```
